```python
import pandas as pd
import numpy as np
df = pd.read_csv('../Datasets/Clear_Dataset.csv')
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
df.head()
```

```python
df = df.rename(columns={
    'Patient ID': 'Patient_ID',
    'Gender (M-male, F-female)': 'Gender',
    'Job (1-normal, 2-intermediate, 3-heavy)': 'Job',
    'Family Background of Diabetes (1-yes, 0-no)': 'Family_History',
    'Height (cm)': 'Height',
    'Diabetic Year': 'Diabetic_Year',
    'Age': 'Age',
    'Weight (kg)': 'Weight',
    'BMI': 'BMI',
    'Follow suggested Diet (1-yes, 0-no)': 'Diet',
    'Take Medicine for Diabetes (1-yes, 0-no)': 'Medicine',
    'Take Insulin (1-yes, 0-no)': 'Insulin',
    'Hypertension (1-yes, 0-no)': 'Hypertension',
    'Heart Disease (1-yes, 0-no)': 'Heart_Disease',
    'Sleep (1-sufficient, 0-insufficient)': 'Sleep',
    'Water Consumption (1-sufficient, 0-insufficient)': 'Water',
    'Smoke (1-yes, 0-no)': 'Smoke',
    'Zarda, Betel Leaf (1-yes, 0-no)': 'Zarda_Betel',
    'Walk Regularly (1-yes, 0-no)': 'Walk',
    'Urination Properly (1-yes, 0-no)': 'Urination',
    'Urinary Infection (1-yes, 0-no)': 'Urinary_Infection',
    'Pain killer (1-yes, 0-no)': 'Pain_Killer',
    'Calorie Intake (per day)': 'Calorie',
    'Next year CKD (1-yes, 0-no)': 'CKD'
})
df = df.drop(['Patient_ID'], axis=1)
df.head()
```

```python
for c in df.columns:
    if not pd.api.types.is_numeric_dtype(df[c]):
        df[c] = df[c].astype(str).str.replace(' ', '', regex=False).str.replace('\t', '', regex=False).str.replace('?', '', regex=False)
        df[c] = df[c].replace('', np.nan)
df.dtypes
```

```python
for c in df.columns:
    print('Number of unique values in', c, ': \n', df[c].nunique())
    print('Values are: \n', sorted(df[c].unique().astype(str)))
```

```python
from sklearn import preprocessing

label_encoder = preprocessing.LabelEncoder()

columns_to_encode = ['Gender']

for column in columns_to_encode:
    values = df[column].dropna()
    codes = pd.Series(label_encoder.fit_transform(values), index=values.index)
    df[column] = codes.reindex(df.index)
df.head()
```

```python
df['CKD'] = pd.to_numeric(df['CKD']).astype(int)
df.head()
```

```python
import os
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import MinMaxScaler

df = pd.read_csv("../Datasets/PreProcessed.csv")

y = df['CKD']
X = df.drop(['CKD'], axis = 1)

columns_with_nulls1 = ['Gender', 'Job', 'Family_History', 'Diet', 'Medicine', 'Insulin', 'Hypertension', 'Heart_Disease', 'Sleep', 'Water', 'Smoke', 'Zarda_Betel', 'Walk', 'Urination', 'Urinary_Infection', 'Pain_Killer']

columns_with_nulls2 = ['Height', 'Diabetic_Year', 'Age', 'Weight', 'BMI', 'Calorie']

split_ratios = [0.8]
```

```python
for ratio in split_ratios:
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=1 - ratio, random_state=42, stratify=y)
    X_train = X_train.copy()
    X_test = X_test.copy()

    imputer1 = SimpleImputer(strategy='most_frequent').fit(X_train[columns_with_nulls1])
    imputer2 = SimpleImputer(strategy='mean').fit(X_train[columns_with_nulls2])
    X_train[columns_with_nulls1] = imputer1.transform(X_train[columns_with_nulls1])
    X_train[columns_with_nulls2] = imputer2.transform(X_train[columns_with_nulls2])
    X_test[columns_with_nulls1] = imputer1.transform(X_test[columns_with_nulls1])
    X_test[columns_with_nulls2] = imputer2.transform(X_test[columns_with_nulls2])

    scaler = MinMaxScaler().fit(X_train)
    train_scaled = pd.DataFrame(scaler.transform(X_train), columns=X.columns)
    test_scaled = pd.DataFrame(scaler.transform(X_test), columns=X.columns)
    train_scaled['CKD'] = y_train.values
    test_scaled['CKD'] = y_test.values

    folder = '../Datasets/Split_' + str(round(ratio * 100)) + '/'
    os.makedirs(folder, exist_ok=True)
    train_scaled.to_csv(folder + 'MinMax_train.csv', index=False)
    test_scaled.to_csv(folder + 'MinMax_test.csv', index=False)
    print(ratio, train_scaled.shape, test_scaled.shape)
```
